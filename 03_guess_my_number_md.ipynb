Guess My Number

I am thinking of a number between 0 and 99... Enter a guess: 50 Your guess is too high

Enter a new number: 25 Your guess is too low

Enter a new number: 40 Your guess is too low

Enter a new number: 45 Your guess is too low

Enter a new number: 48 Congrats! The number was: 48

```python
import random

def main():
    # Generate the secret number at random between 1 and 99
    secret_number = random.randint(1, 99)

    print("I am thinking of a number between 1 and 99...")

    # First guess from user
    guess = int(input("Enter a guess: "))

    # Keep looping until the guess is correct
    while guess != secret_number:
        if guess < secret_number:
            print("Your guess is too low")
        else:
            print("Your guess is too high")

        print()  # Adds a blank line for cleaner output
        guess = int(input("Enter a new guess: "))

    # If we break out of the loop, the guess was correct
    print("Congrats! The number was:", secret_number)

# Run the game
if __name__ == '__main__':
    main()
```
